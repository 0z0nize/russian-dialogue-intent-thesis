# Создание структуры папок на Google Drive

Этот ноутбук создаёт необходимую структуру папок проекта на Google Drive.

Запустите все ячейки по порядку в Google Colab.

## 1. Подключение Google Drive

```python
from google.colab import drive
drive.mount('/content/drive')
```

## 2. Создание папок

```python
import os

BASE_DIR = "/content/drive/MyDrive/russian-dialogue-intent-thesis"

FOLDERS = [
    "data/raw",
    "data/interim",
    "data/processed",
    "data/annotation",
    "models",
    "results/tables",
    "results/figures",
    "results/logs",
    "results/predictions",
    "exports",
]

for folder in FOLDERS:
    path = os.path.join(BASE_DIR, folder)
    os.makedirs(path, exist_ok=True)
    print(f"Создана папка: {path}")
```

## 3. Проверка структуры

```python
for folder in FOLDERS:
    path = os.path.join(BASE_DIR, folder)
    exists = os.path.isdir(path)
    print(f"{'OK' if exists else 'НЕТ'}: {path}")
```
